## 6. Seasonal Agricultural Performance

The agricultural performance of different seasons is compared using average yield, profit, water efficiency, and disease or pest risk.


```python
season_performance = df.groupby('Season').agg({
    'Yield_Tonnes_Ha': 'mean',
    'Profit_INR': 'mean',
    'Water_Efficiency_t_per_1000m3': 'mean',
    'Disease_Pest_Risk_pct': 'mean'
}).reset_index()

season_performance
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=season_performance, x='Season', y='Yield_Tonnes_Ha', palette='viridis')
plt.title('Average Crop Yield Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=season_performance, x='Season', y='Profit_INR', palette='magma')
plt.title('Average Profit Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=season_performance, x='Season', y='Water_Efficiency_t_per_1000m3', palette='crest')
plt.title('Average Water Efficiency Across Seasons')
plt.xlabel('Season')
plt.ylabel('Water Efficiency (Tonnes per 1000 m³)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=season_performance, x='Season', y='Disease_Pest_Risk_pct', palette='rocket')
plt.title('Average Disease and Pest Risk Across Seasons')
plt.xlabel('Season')
plt.ylabel('Risk (%)')
plt.tight_layout()
plt.show()
```

## 7. Crop × Season Analysis

The performance of different crops is compared across Kharif, Rabi, and Zaid seasons. Average yield, profit, and water efficiency are analyzed to identify the best-performing crop-season combinations.


```python
crop_season_yield = df.pivot_table(
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

plt.figure(figsize=(10, 6))
sns.heatmap(crop_season_yield, annot=True, fmt='.2f', cmap='YlGnBu')
plt.title('Average Yield by Crop and Season')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.tight_layout()
plt.show()
```

```python
crop_season_profit = df.pivot_table(
    values='Profit_INR',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

plt.figure(figsize=(10, 6))
sns.heatmap(crop_season_profit, annot=True, fmt='.0f', cmap='RdYlGn', center=0)
plt.title('Average Profit by Crop and Season')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.tight_layout()
plt.show()
```

```python
crop_season_water = df.pivot_table(
    values='Water_Efficiency_t_per_1000m3',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

plt.figure(figsize=(10, 6))
sns.heatmap(crop_season_water, annot=True, fmt='.2f', cmap='PuBuGn')
plt.title('Average Water Efficiency by Crop and Season')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.tight_layout()
plt.show()
```

## 9. Statistical Correlation Analysis

Correlation analysis is used to measure the strength and direction of relationships between important agricultural variables. This helps identify which environmental and operational factors are associated with yield, profit, water efficiency, and disease or pest risk.


```python
analysis_columns = [
    'Farm_Area_Hectares',
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Soil_Moisture_pct',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Market_Price_INR_Tonne',
    'Total_Cost_INR',
    'Revenue_INR',
    'Profit_INR',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3',
    'Disease_Pest_Risk_pct'
]

correlation_matrix = df[analysis_columns].corr()

plt.figure(figsize=(14, 10))
sns.heatmap(
    correlation_matrix,
    cmap='RdBu_r',
    center=0,
    annot=True,
    fmt='.2f'
)
plt.title('Correlation Matrix of Agricultural Variables')
plt.tight_layout()
plt.show()
```

### Factors Associated with Profit

The correlation of different agricultural variables with profit is examined to identify the variables most strongly associated with economic performance.


```python
profit_correlation = correlation_matrix['Profit_INR'].drop('Profit_INR').sort_values()

plt.figure(figsize=(10, 7))
profit_correlation.plot(kind='barh', color='teal')
plt.title('Correlation of Agricultural Variables with Profit')
plt.xlabel('Correlation with Profit')
plt.ylabel('Agricultural Variable')
plt.tight_layout()
plt.show()
```

## 10. Reliability Analysis

Reliability analysis is used to identify crop-season combinations that provide consistent agricultural and economic performance. Yield, profit, water efficiency, disease or pest risk, and profitability across states are considered to compare the reliability of different combinations.


```python
reliability = df.groupby(['Crop', 'Season']).agg(
    Median_Profit=('Profit_INR', 'median'),
    Avg_Yield=('Yield_Tonnes_Ha', 'mean'),
    Avg_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean'),
    Avg_Risk=('Disease_Pest_Risk_pct', 'mean'),
    Profitable_Records=('Profit_INR', lambda x: (x > 0).mean() * 100)
).reset_index()

reliability['Reliability_Score'] = (
    reliability['Profitable_Records'].rank(pct=True) * 40 +
    reliability['Avg_Yield'].rank(pct=True) * 25 +
    reliability['Avg_Water_Efficiency'].rank(pct=True) * 20 +
    (1 - reliability['Avg_Risk'].rank(pct=True)) * 15
)

reliability = reliability.sort_values('Reliability_Score', ascending=False)

reliability
```

```python
top_reliability = reliability.head(10).copy()

top_reliability['Crop_Season'] = (
    top_reliability['Crop'] + ' - ' + top_reliability['Season']
)

plt.figure(figsize=(12, 7))
sns.barplot(
    data=top_reliability,
    x='Reliability_Score',
    y='Crop_Season',
    hue='Crop_Season',
    palette='viridis',
    legend=False
)
plt.title('Top 10 Most Reliable Crop-Season Combinations')
plt.xlabel('Reliability Score')
plt.ylabel('Crop - Season')
plt.tight_layout()
plt.show()
```

## 11. Regional Consistency Analysis

Regional consistency is analyzed by comparing the profitability of crop-season combinations across different states. This helps identify combinations that perform consistently across regions and combinations that show greater variation in economic outcomes.


```python
regional_consistency = df.groupby(['Crop', 'Season']).agg(
    States=('State', 'nunique'),
    Profitable_States=('Profit_INR', lambda x: (x > 0).sum()),
    Avg_Profit=('Profit_INR', 'mean')
).reset_index()

regional_consistency['Profitability_Percentage'] = (
    regional_consistency['Profitable_States'] /
    regional_consistency['States'] * 100
)

regional_consistency.sort_values(
    'Profitability_Percentage',
    ascending=False
).head(15)
```

```python
top_regional = regional_consistency.sort_values(
    'Profitability_Percentage',
    ascending=False
).head(10).copy()

top_regional['Crop_Season'] = (
    top_regional['Crop'] + ' - ' + top_regional['Season']
)

plt.figure(figsize=(12, 7))
sns.barplot(
    data=top_regional,
    x='Profitability_Percentage',
    y='Crop_Season',
    hue='Crop_Season',
    palette='Set2',
    legend=False
)
plt.title('Regional Profitability Consistency')
plt.xlabel('Profitable Records (%)')
plt.ylabel('Crop - Season')
plt.xlim(0, 100)
plt.tight_layout()
plt.show()
```
